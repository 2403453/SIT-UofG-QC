```python
%pip install qiskit==1.2.4
%pip install qiskit-aer==0.15.1
%pip install pylatexenc==2.10

from qiskit import QuantumCircuit
from qiskit.converters import circuit_to_gate
from qiskit.visualization import array_to_latex
from qiskit.quantum_info import Operator
from qiskit import transpile
from qiskit.providers.basic_provider import BasicSimulator
from qiskit.providers.fake_provider import GenericBackendV2
from qiskit.visualization import plot_histogram
from qiskit.result import marginal_counts
```

```python
# Teleportation
# Create a circuit with 3 qubits (first parameter) and 3 classical bits (second parameter)
teleport = QuantumCircuit(3,3)

# qubit z is the state to be teleported from Alice
# qubits x and y are the shared entangled state
# Alice has qubit x, Bob has qubit y
# z -> left qubit     2
# x -> middle qubit   1
# y -> right qubit    0


# Add gates here to entangle qubits x and y
teleport.h(1)
teleport.cx(1,0)

# Add gates here to entangle qubit z with the others
teleport.cx(2,1)
teleport.h(2)

# Measure qubits 2 and 1 and put the results on classical bits 2 and 1.
teleport.measure([2,1],[2,1])

# Fix up the final state of qubit y, depending on the measurement results.
# Here we are using a feature of Qiskit that we haven't seen before.
with teleport.if_test((1, 1)):
    teleport.x(0)
with teleport.if_test((2, 1)):
    teleport.z(0)

# Sometimes it's useful to insert a "barrier" to make the diagram clearer.
# A barrier also has an effect on how the circuit is compiled, but we won't worry about that.
teleport.barrier()

# Measure qubit y (first parameter) and put the result on classical bit 0 (second parameter)
teleport.measure([0],[0])

# Check the circuit
teleport.draw("mpl")
```

```python
# Use a circuit to put qubit z into the state that we want to teleport,
# and compose it into the main circuit.

prep = QuantumCircuit(1)
# Add a gate here to put qubit z into some state other than |0>
prep.x(0)

# When we compose circuits, the second parameter is the list of qubits (wires) that are being connected
teleport = teleport.compose(prep, [2], front=True)
teleport.draw("mpl")
```

```python
# Now simulate and plot the results.
# In general this won't tell us that we have exactly the right state,
# just that we get the right measurement statistics.
# We are interested in the result on classical bit 2, which is the most significant bit (leftmost).
# The simulator adds some noise, so we get some non-zero probabilities for parts of bit 2 that shouldn't be there.
# To see results without noise, add the parameter noise_info=False to GenericBackendV2
# We need to use GenericBackendV2 instead of BasicSimulator, and set control_flow=True, so that we can use if_test.

backend = GenericBackendV2(num_qubits=3, control_flow=True)
compiled = transpile(teleport, backend)
job_sim = backend.run(compiled, shots=1024)
result_sim = job_sim.result()
counts = result_sim.get_counts(compiled)
mcounts = marginal_counts(counts,indices=[0]) # We only want to see the results for bit 0
plot_histogram(mcounts)
```

```python
# Teleportation has an interesting feature which is not usually
# included in the description.

# If the qubit being teleported is entangled with other qubits, then after
# teleportation it is still entangled in the same way.

# Try this out!
# You will need to introduce another qubit and entangle it with qubit 0.
# After teleporting, work out how to use measurements to convince yourself that qubit 2 is entangled with the extra qubit.
# You can do this in a similar way to the exercise in Lab1 about Bell states.

# Teleportation
# Create a circuit with 4 qubits (first parameter) and 4 classical bits (second parameter)
teleport = QuantumCircuit(4,4)

# qubit z is the state to be teleported from Alice
# qubits x and y are the shared entangled state
# Alice has qubit x, Bob has qubit y
# z -> left qubit     3
# x -> middle qubit   2
# y -> right qubit    1
# k -> extra entangled bit 0


# Add gates here to entangle qubits x and y
teleport.h(2)
teleport.cx(2,1)

# Add gates here to entangle qubit z with the others
teleport.cx(3,2)
teleport.h(3)

# Measure qubits 3 and 2 and put the results on classical bits 3 and 2.
teleport.measure([3,2],[3,2])

# Fix up the final state of qubit y, depending on the measurement results.
# Here we are using a feature of Qiskit that we haven't seen before.
with teleport.if_test((2, 1)):
    teleport.x(1)
with teleport.if_test((3, 1)):
    teleport.z(1)

# Sometimes it's useful to insert a "barrier" to make the diagram clearer.
# A barrier also has an effect on how the circuit is compiled, but we won't worry about that.
teleport.barrier()

# Measure qubit y and k (first parameter) and put the result on classical bit 1 and 0 (second parameter)
teleport.measure([1,0],[1,0])

# Check the circuit
teleport.draw("mpl")
```

```python
# Use a circuit to put qubits 0 and 1 into a Bell state,
# and compose it into the main circuit.

prep = QuantumCircuit(4)
prep.h(3)
prep.cx(3,0)

# Barrier to make the diagram clearer
prep.barrier()

teleport = teleport.compose(prep, [0,1,2,3], front=True)
teleport.draw("mpl")
```

```python
# Now simulate and plot the results. We are interested in classical bits 2 and 3
# In general this won't tell us that we have exactly the right state,
# just that we get the right measurement statistics.
# The simulator adds some noise, so we get some non-zero probabilities for parts of bit 2 that shouldn't be there.
# To see results without noise, add the parameter noise_info=False to GenericBackendV2

backend = GenericBackendV2(num_qubits=4, control_flow=True)
compiled = transpile(teleport, backend)
job_sim = backend.run(compiled, shots=1024)
result_sim = job_sim.result()
counts = result_sim.get_counts(compiled)
mcounts = marginal_counts(counts,indices=[0,1]) # We only want to see the results for bits 2 and 3
plot_histogram(mcounts)
```
